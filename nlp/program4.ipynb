```
import numpy as np
import pandas as pd
import nltk
nltk.download('punkt')
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```
# Sample Dataset (or use pd.read_csv for your own dataset)
# df = pd.read_csv('your_data.csv')
# X = df['text_column'].tolist()
# y = df['label_column'].tolist()
X = ["great product", "best service", "love it", "excellent", "bad", "terrible", "waste", "poor"]
y = [1, 1, 1, 1, 0, 0, 0, 0]
```

```
def preprocess(t):
    return [w for w in word_tokenize(t.lower()) if w.isalnum() and w not in set(stopwords.words('english'))]
```

```
def get_ngrams(words, n=2):
    return [tuple(words[i:i+n]) for i in range(len(words)-n+1)]
```

```
processed = [get_ngrams(preprocess(s)) for s in X]
vocab = list(set(ng for d in processed for ng in d))
X_train = np.array([[d.count(ng) for ng in vocab] for d in processed])
```

```
class LogisticRegression:
    def __init__(self, lr=0.1, iters=1000):
        self.lr, self.iters = lr, iters
    def fit(self, X, y):
        self.w = np.zeros(X.shape[1])
        self.b = 0
        for _ in range(self.iters):
            a = 1/(1+np.exp(-(X.dot(self.w)+self.b)))
            self.w -= self.lr*(X.T.dot(a-y))/len(y)
            self.b -= self.lr*(a-y).sum()/len(y)
    def predict(self, X):
        return (1/(1+np.exp(-(X.dot(self.w)+self.b)))>=0.5).astype(int)
```

```
model = LogisticRegression()
model.fit(X_train, y)
test = get_ngrams(preprocess("best service"))
test_vec = np.array([[test.count(ng) for ng in vocab]])
print("Prediction:", model.predict(test_vec)[0])
```
