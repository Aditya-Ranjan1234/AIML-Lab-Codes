```
import numpy as np
import pandas as pd
import nltk
nltk.download('punkt')
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```
# Sample Dataset (or use pd.read_csv for your own dataset)
# df = pd.read_csv('your_data.csv')
# X = df['text_column'].tolist()
# y = df['label_column'].tolist()
X = ["great product", "best service", "love it", "excellent", "bad", "terrible", "waste", "poor"]
y = [1, 1, 1, 1, 0, 0, 0, 0]
```

```
def preprocess(t):
    return [w for w in word_tokenize(t.lower()) if w.isalnum() and w not in set(stopwords.words('english'))]
```

```
processed = [preprocess(s) for s in X]
vocab = list(set(w for d in processed for w in d))
X_train = np.array([[d.count(w) for w in vocab] for d in processed])
```

```
class NaiveBayes:
    def fit(self, X, y):
        self.classes = np.unique(y)
        self.priors = {c: np.mean(y == c) for c in self.classes}
        self.likelihoods = {}
        for c in self.classes:
            Xc = X[y == c]
            self.likelihoods[c] = (Xc.sum(axis=0) + 1) / (Xc.sum() + len(vocab))
    def predict(self, X):
        preds = []
        for x in X:
            post = {}
            for c in self.classes:
                post[c] = np.log(self.priors[c]) + np.sum(np.log(self.likelihoods[c]) * x)
            preds.append(max(post, key=post.get))
        return preds
```

```
model = NaiveBayes()
model.fit(X_train, y)

test = preprocess("great product")
test_vec = np.array([[test.count(w) for w in vocab]])
print("Prediction:", model.predict(test_vec)[0])
```
