```python
import numpy as np
import nltk
nltk.download('punkt')
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
X = ["great product", "best service", "love it", "excellent", "bad", "terrible", "waste", "poor"]
y = [1, 1, 1, 1, 0, 0, 0, 0]
```

```python
def preprocess(text):
    stop_words = set(stopwords.words('english'))
    return [w for w in word_tokenize(text.lower()) if w.isalnum() and w not in stop_words]
```

```python
processed = [preprocess(s) for s in X]
vocab = list(set(w for d in processed for w in d))
X_train = np.array([[d.count(w) for w in vocab] for d in processed])
```

```python
class NaiveBayes:
    def fit(self, X, y):
        self.classes = np.unique(y)
        self.priors = {c: np.mean(y == c) for c in self.classes}
        self.likelihoods = {}
        for c in self.classes:
            X_c = X[y == c]
            self.likelihoods[c] = (X_c.sum(axis=0) + 1) / (X_c.sum() + len(vocab))
    def predict(self, X):
        preds = []
        for x in X:
            posteriors = {}
            for c in self.classes:
                prior = np.log(self.priors[c])
                likelihood = np.sum(np.log(self.likelihoods[c]) * x)
                posteriors[c] = prior + likelihood
            preds.append(max(posteriors, key=posteriors.get))
        return preds
```

```python
model = NaiveBayes()
model.fit(X_train, y)
```

```python
test = preprocess("great product")
test_vec = np.array([[test.count(w) for w in vocab]])
print("Prediction:", model.predict(test_vec)[0])
```
