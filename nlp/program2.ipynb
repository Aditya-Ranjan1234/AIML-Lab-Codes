```
import matplotlib.pyplot as plt
import random
import nltk
import pandas as pd
nltk.download('punkt')
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```
# Sample Dataset (or use pd.read_csv for your own dataset)
# df = pd.read_csv('your_data.csv')
# text = ' '.join(df['text_column'].tolist())
text = "Tendulkar made his debut in 1989. He scored many centuries. He is a legend."
```

```
def get_freq(t):
    freq = {}
    for w in word_tokenize(t.lower()):
        if w.isalnum() and w not in set(stopwords.words('english')):
            freq[w] = freq.get(w, 0) + 1
    return sorted(freq.items(), key=lambda x: x[1], reverse=True)
```

```
def word_cloud(t):
    freq = get_freq(t)
    grid = set()
    plt.figure(figsize=(8, 8))
    for word, count in freq[:30]:
        while True:
            x = random.randint(0, 7)
            y = random.randint(0, 7)
            if (x, y) not in grid:
                grid.add((x, y))
                break
        plt.text(x/8, y/8, word, fontsize=10 + count*3, color=(random.random(), random.random(), random.random()), ha='center')
    plt.axis('off')
    plt.show()
```

```
word_cloud(text)
```
