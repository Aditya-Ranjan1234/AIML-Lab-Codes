```
import numpy as np
import pandas as pd
import nltk
nltk.download('punkt')
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```
# Sample Dataset (or use pd.read_csv for your own dataset)
# df = pd.read_csv('your_data.csv')
# X = df['text_column'].tolist()
# y = df['label_column'].tolist()
X = ["great product", "best service", "love it", "excellent", "bad", "terrible", "waste", "poor"]
y = [1, 1, 1, 1, 0, 0, 0, 0]
```

```
def preprocess(t):
    return [w for w in word_tokenize(t.lower()) if w.isalnum() and w not in set(stopwords.words('english'))]
```

```
processed = [preprocess(s) for s in X]
vocab = list(set(w for d in processed for w in d))
X_train = []
for d in processed:
    vec = []
    for w in vocab:
        tf = d.count(w) / (len(d) + 1)
        idf = np.log(len(processed) / (sum(1 for doc in processed if w in doc) + 1))
        vec.append(tf * idf)
    X_train.append(vec)
X_train = np.array(X_train)
```

```
class LogisticRegression:
    def __init__(self, lr=0.1, iters=1000):
        self.lr = lr
        self.iters = iters
    def fit(self, X, y):
        self.w = np.zeros(X.shape[1])
        self.b = 0
        for _ in range(self.iters):
            z = X.dot(self.w) + self.b
            a = 1 / (1 + np.exp(-z))
            self.w -= self.lr * (X.T.dot(a - y)) / len(y)
            self.b -= self.lr * (a - y).sum() / len(y)
    def predict(self, X):
        return (1 / (1 + np.exp(-(X.dot(self.w) + self.b))) >= 0.5).astype(int)
```

```
model = LogisticRegression()
model.fit(X_train, y)
test = preprocess("great service")
test_vec = []
for w in vocab:
    tf = test.count(w) / (len(test) + 1)
    idf = np.log(len(processed) / (sum(1 for doc in processed if w in doc) + 1))
    test_vec.append(tf * idf)
test_vec = np.array([test_vec])
print("Prediction:", model.predict(test_vec)[0])
```
