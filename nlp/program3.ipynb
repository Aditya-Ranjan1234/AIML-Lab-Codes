```python
import numpy as np
import nltk
nltk.download('punkt')
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
X = ["great product", "best service", "love it", "excellent", "bad", "terrible", "waste", "poor"]
y = [1, 1, 1, 1, 0, 0, 0, 0]
```

```python
def preprocess(text):
    stop_words = set(stopwords.words('english'))
    return [w for w in word_tokenize(text.lower()) if w.isalnum() and w not in stop_words]
```

```python
def tfidf(word, doc, docs):
    tf = doc.count(word) / (len(doc) + 1)
    idf = np.log(len(docs) / (sum(1 for d in docs if word in d) + 1))
    return tf * idf
```

```python
processed = [preprocess(s) for s in X]
vocab = list(set(w for d in processed for w in d))
X_train = np.array([[tfidf(w, d, processed) for w in vocab] for d in processed])
```

```python
class LogisticRegression:
    def __init__(self, lr=0.1, iters=1000):
        self.lr = lr
        self.iters = iters
    def sigmoid(self, z):
        return 1 / (1 + np.exp(-z))
    def fit(self, X, y):
        self.w = np.zeros(X.shape[1])
        self.b = 0
        for _ in range(self.iters):
            z = X.dot(self.w) + self.b
            a = self.sigmoid(z)
            dw = (X.T.dot(a - y)) / len(y)
            db = (a - y).sum() / len(y)
            self.w -= self.lr * dw
            self.b -= self.lr * db
    def predict(self, X):
        return (self.sigmoid(X.dot(self.w) + self.b) >= 0.5).astype(int)
```

```python
model = LogisticRegression()
model.fit(X_train, y)
```

```python
test = preprocess("great service")
test_vec = np.array([[tfidf(w, test, processed) for w in vocab]])
print("Prediction:", model.predict(test_vec)[0])
```
