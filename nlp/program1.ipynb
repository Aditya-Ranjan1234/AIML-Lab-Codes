```
import nltk, numpy as np
nltk.download('punkt');nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
def tf(w,s):return word_tokenize(s.lower()).count(w)/len(word_tokenize(s.lower()))
def idf(w,ss):return np.log(len(ss)/(sum(1 for s in ss if w in word_tokenize(s.lower()))+1))
def pre(t):sw=set(stopwords.words('english'));return [w for w in word_tokenize(t.lower()) if w.isalnum() and w not in sw]
def sumr(t,n=3):ss=sent_tokenize(t);sc=[(sum(tf(w,s)*idf(w,ss) for w in pre(s)),s) for s in ss];return ' '.join([s for _,s in sorted(sc,key=lambda x:x[0],reverse=True)[:n]])
text="""Dhoni made his debut in 2004. He became captain in 2007. He retired in 2019."""
print(sumr(text))
```
