```
import nltk
import numpy as np
import pandas as pd
nltk.download('punkt')
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
```

```
def preprocess(text):
    sw = set(stopwords.words('english'))
    return [w for w in word_tokenize(text.lower()) if w.isalnum() and w not in sw]
```

```
def tf(word, sent): return word_tokenize(sent.lower()).count(word)/(len(word_tokenize(sent.lower()))+1)
```

```
def idf(word, sents):
    count = sum(1 for s in sents if word in word_tokenize(s.lower()))
    return np.log(len(sents)/(count+1))
```

```
def summarize(text, n=3):
    sents = sent_tokenize(text)
    scores = []
    for sent in sents:
        words = preprocess(sent)
        score = sum(tf(w, sent)*idf(w, sents) for w in words)
        scores.append((score, sent))
    top = sorted(scores, key=lambda x:x[0], reverse=True)[:n]
    return ' '.join([s for _,s in top)
```
