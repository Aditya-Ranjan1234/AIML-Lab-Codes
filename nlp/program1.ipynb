```
import nltk
import numpy as np
import pandas as pd
nltk.download('punkt')
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
```

```
# Sample Dataset (or use pd.read_csv for your own dataset)
# df = pd.read_csv('your_data.csv')
# text = df['text_column'].iloc[0]
text = "Dhoni made his debut in 2004. He became captain in 2007. He retired in 2019. He is a legend."
```

```
def preprocess(t):
    return [w for w in word_tokenize(t.lower()) if w.isalnum() and w not in set(stopwords.words('english'))]
```

```
def summarize(t, n=3):
    sents = sent_tokenize(t)
    scores = []
    for s in sents:
        words = preprocess(s)
        words = word_tokenize(s.lower())
        tf = sum([words.count(w) / (len(words) + 1) for w in preprocess(s)])
        scores.append((tf, s))
    top = sorted(scores, key=lambda x: x[0], reverse=True)[:n]
    return ' '.join([s for _, s in top])
```

```
print("Summary:", summarize(text))
```
